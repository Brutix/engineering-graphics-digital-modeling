# Інтерактивна побудова спряження

Змінюйте **радіус R** та **кут між прямими** повзунками. Геометрія перебудовується автоматично.

**Ідея:** параметри → алгоритм → графічний результат.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider


# ============================================================
# АЛГОРИТМ ПОБУДОВИ СПРЯЖЕННЯ
# ============================================================

def draw_fillet(R=2.0, ANGLE=60):
    angle = np.radians(ANGLE)

    # Вершина кута
    B = np.array([0.0, 0.0])

    # Напрямки двох прямих
    d1 = np.array([-1.0, 0.0])
    d2 = np.array([np.cos(angle), np.sin(angle)])

    # Відстань від вершини до точок дотику
    t = R / np.tan(angle / 2)

    # Точки дотику
    P1 = B + d1 * t
    P2 = B + d2 * t

    # Центр дуги спряження
    O = np.array([-t, R])

    # Довжини прямих для відображення
    L1 = max(6, t + 2)
    L2 = max(6, t + 2)

    A = B + d1 * L1
    C = B + d2 * L2

    # Кути початку та кінця дуги
    a1 = np.arctan2(P1[1] - O[1], P1[0] - O[0])
    a2 = np.arctan2(P2[1] - O[1], P2[0] - O[0])

    # Точки дуги
    arc_angles = np.linspace(a1, a2, 150)
    x_arc = O[0] + R * np.cos(arc_angles)
    y_arc = O[1] + R * np.sin(arc_angles)

    # ------------------------------------------------------------
    # ВІЗУАЛІЗАЦІЯ
    # ------------------------------------------------------------
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # Початкові прямі
    ax.plot([A[0], B[0]], [A[1], B[1]], linewidth=2, label="Пряма 1")
    ax.plot([B[0], C[0]], [B[1], C[1]], linewidth=2, label="Пряма 2")

    # Дуга спряження
    ax.plot(x_arc, y_arc, linewidth=4, label=f"Спряження R = {R:.1f}")

    # Точки дотику
    ax.scatter([P1[0], P2[0]], [P1[1], P2[1]], s=65, zorder=5,
               label="Точки дотику")

    # Центр дуги
    ax.scatter(O[0], O[1], s=65, zorder=5, label="Центр дуги")

    # Радіуси до точок дотику
    ax.plot([O[0], P1[0]], [O[1], P1[1]], linestyle="--", linewidth=1.2)
    ax.plot([O[0], P2[0]], [O[1], P2[1]], linestyle="--", linewidth=1.2)

    # Підписи
    ax.text(P1[0], P1[1] - 0.35, "P₁", fontsize=12)
    ax.text(P2[0] + 0.12, P2[1], "P₂", fontsize=12)
    ax.text(O[0] + 0.12, O[1] + 0.12, "O", fontsize=12)
    ax.text(B[0] + 0.12, B[1] - 0.35, "B", fontsize=12)

    ax.set_aspect("equal", adjustable="box")
    ax.grid(True, alpha=0.25)
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_title(f"Алгоритмічна побудова спряження | R = {R:.1f} | кут = {ANGLE}°")
    ax.legend(loc="best")
    plt.show()


# ============================================================
# КЕРУВАННЯ — СТУДЕНТ ЗМІНЮЄ ТІЛЬКИ ЦІ ПАРАМЕТРИ
# ============================================================

interact(
    draw_fillet,
    R=FloatSlider(
        value=2.0, min=0.5, max=5.0, step=0.1,
        description="Радіус R:", continuous_update=True
    ),
    ANGLE=IntSlider(
        value=60, min=20, max=160, step=5,
        description="Кут:", continuous_update=True
    )
);
```

## Що спостерігає студент

- **R** збільшує або зменшує радіус спряження.
- **Кут** змінює геометрію вихідного кута.
- Точки **P₁** і **P₂** автоматично переміщуються.
- Центр **O** автоматично перераховується.
- Дуга перебудовується без ручного креслення.

Це демонструє принцип: **геометричний параметр → математичний алгоритм → цифрове відображення**.
