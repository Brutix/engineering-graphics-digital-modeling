```python
import plotly.graph_objects as go
from PIL import Image

def create_a4_frame_with_drawing(
    doc_name="ВТУЛКА ЗУБЧАСТА\n(СПЕЦІАЛЬНА)",
    doc_code="КПІ.140224.004 СП",
    author="студент",
    checker="Коваленко О.П.",
    approver="Шевченко В.М.",
    scale="2:1",
    weight="1.2 кг",
    sheet_num="1",
    sheets_count="1",
    material="Сталь 40Х ГОСТ 4543-71",
    company="НТУУ \"КПІ\""
):
    fig = go.Figure()

    # --- 1. Зовнішня межа аркуша A4 (210 x 297 мм) ---
    fig.add_shape(
        type="rect",
        x0=0, y0=0, x1=210, y1=297,
        line=dict(color="#CCCCCC", width=1, dash="dot"),
        fillcolor="white"
    )

    # --- 2. Внутрішня рамка креслення (2.5 мм) ---
    # Робоче поле: X ∈ [20, 205], Y ∈ [5, 292]
    fig.add_shape(
        type="rect",
        x0=20, y0=5, x1=205, y1=292,
        line=dict(color="black", width=2.5),
        fillcolor="rgba(0,0,0,0)"
    )

    # --- 3. Контур та геометрія штампа (185 х 55 мм) ---
    x_stamp = 20
    y_stamp = 5

    fig.add_shape(
        type="rect",
        x0=x_stamp, y0=y_stamp, x1=205, y1=60,
        line=dict(color="black", width=2.5),
        fillcolor="white"
    )

    # Горизонтальні лінії штампа
    left_h_lines = [(10, 2.5), (15, 1), (20, 1), (25, 1), (30, 1)]
    for y_off, w in left_h_lines:
        fig.add_shape(
            type="line",
            x0=x_stamp, y0=y_stamp + y_off,
            x1=x_stamp + 65, y1=y_stamp + y_off,
            line=dict(color="black", width=w)
        )

    fig.add_shape(
        type="line",
        x0=x_stamp, y0=y_stamp + 35,
        x1=205, y1=y_stamp + 35,
        line=dict(color="black", width=2.5)
    )

    fig.add_shape(
        type="line",
        x0=x_stamp + 65, y0=y_stamp + 7,
        x1=x_stamp + 135, y1=y_stamp + 7,
        line=dict(color="black", width=1)
    )

    right_h_lines = [(7, 1), (15, 1), (20, 1)]
    for y_off, w in right_h_lines:
        fig.add_shape(
            type="line",
            x0=x_stamp + 135, y0=y_stamp + y_off,
            x1=205, y1=y_stamp + y_off,
            line=dict(color="black", width=w)
        )

    # Вертикальні лінії штампа
    v_offsets_left = [7, 17, 40, 55, 65]
    for x_off in v_offsets_left:
        fig.add_shape(
            type="line",
            x0=x_stamp + x_off, y0=y_stamp,
            x1=x_stamp + x_off, y1=y_stamp + 35,
            line=dict(color="black", width=1)
        )

    fig.add_shape(
        type="line",
        x0=x_stamp + 135, y0=y_stamp,
        x1=x_stamp + 135, y1=y_stamp + 35,
        line=dict(color="black", width=1)
    )

    fig.add_shape(
        type="line",
        x0=x_stamp + 150, y0=y_stamp + 15,
        x1=x_stamp + 150, y1=y_stamp + 35,
        line=dict(color="black", width=1)
    )
    fig.add_shape(
        type="line",
        x0=x_stamp + 167, y0=y_stamp + 15,
        x1=x_stamp + 167, y1=y_stamp + 35,
        line=dict(color="black", width=1)
    )
    fig.add_shape(
        type="line",
        x0=x_stamp + 155, y0=y_stamp,
        x1=x_stamp + 155, y1=y_stamp + 15,
        line=dict(color="black", width=1)
    )

    # --- 4. ТЕКСТ ШТАМПА ---
    labels = [
        (x_stamp + 12, y_stamp + 32.5, "Розроб."),
        (x_stamp + 12, y_stamp + 27.5, "Перев."),
        (x_stamp + 12, y_stamp + 17.5, "Н.контр."),
        (x_stamp + 12, y_stamp + 12.5, "Затв."),
        (x_stamp + 142.5, y_stamp + 31, "Літ."),
        (x_stamp + 158.5, y_stamp + 31, "Маса"),
        (x_stamp + 176, y_stamp + 31, "Масштаб"),
        (x_stamp + 145, y_stamp + 11, "Аркуш"),
        (x_stamp + 180, y_stamp + 11, "Аркушів"),
    ]
    
    for x, y, txt in labels:
        fig.add_annotation(
            x=x, y=y, text=txt, showarrow=False,
            font=dict(size=8, color="black", family="Arial"),
            xanchor="center", yanchor="middle"
        )

    doc_name_formatted = doc_name.replace("\n", "<br>")

    dynamic_texts = [
        (x_stamp + 100, y_stamp + 45, doc_code, 13, True, "center"),
        (x_stamp + 100, y_stamp + 21, doc_name_formatted, 11, True, "center"),
        (x_stamp + 100, y_stamp + 3.5, material, 8.5, False, "center"),
        (x_stamp + 28.5, y_stamp + 32.5, author, 8, False, "center"),
        (x_stamp + 28.5, y_stamp + 27.5, checker, 8, False, "center"),
        (x_stamp + 28.5, y_stamp + 12.5, approver, 8, False, "center"),
        (x_stamp + 158.5, y_stamp + 17, weight, 9, False, "center"),
        (x_stamp + 175, y_stamp + 17, scale, 9, False, "center"),
        (x_stamp + 145, y_stamp + 3.5, sheet_num, 9, False, "center"),
        (x_stamp + 180, y_stamp + 3.5, sheets_count, 9, False, "center"),
        (x_stamp + 180, y_stamp + 11, company, 9, True, "center"),
    ]

    for x, y, txt, size, is_bold, align in dynamic_texts:
        fig.add_annotation(
            x=x, y=y, text=f"<b>{txt}</b>" if is_bold else txt,
            showarrow=False,
            font=dict(size=size, color="black", family="Arial"),
            xanchor=align, yanchor="middle"
        )

    # =========================================================================
    # --- 5. РОБОЧА ЗОНА ДЛЯ КРЕСЛЕННЯ ТА ЗОБРАЖЕННЯ ---
    # =========================================================================

    # 5.1 Пунктирна рамка-орієнтир для візуалізації робочої зони
    # Координати робочої зони: X від 25 до 200, Y від 65 до 287
    fig.add_shape(
        type="rect",
        x0=25, y0=65, x1=200, y1=287,
        line=dict(color="red", width=1, dash="dash"),
        fillcolor="rgba(240, 248, 255, 0.3)"  # Легка блакитна підкладка
    )
    
    fig.add_annotation(
        x=112.5, y=280,
        text="<b>РОБОЧА ЗОНА КРЕСЛЕННЯ (X: 20-205 мм, Y: 60-292 мм)</b>",
        showarrow=False,
        font=dict(size=10, color="blue", family="Arial")
    )

    # -------------------------------------------------------------------------
    # ВАРИАНТ А: Додавання векторної графіки / математичних кривих (Trace)
    # Наприклад: побудова контуру деталі або траєкторії руху точки
    # -------------------------------------------------------------------------
    sample_x = [40, 100, 100, 160, 160, 40, 40]
    sample_y = [100, 100, 180, 180, 220, 220, 100]
    
    fig.add_trace(go.Scatter(
        x=sample_x,
        y=sample_y,
        mode='lines+markers',
        name='Контур деталі',
        line=dict(color='darkred', width=2),
        marker=dict(size=6, color='red')
    ))

    # -------------------------------------------------------------------------
    # ВАРІАНТ Б: Вставка зовнішнього PNG / JPEG зображення (Image)
    # Для вставки розкоментуйте код нижче та вкажіть джерело (URL або Base64)
    # -------------------------------------------------------------------------
    # img = Image.open("https://drive.google.com/open?id=1g-T_tZ1rxaLA94z4p9oDLAgdCxlBw1gy&usp=drive_fs")
    # fig.add_layout_image(
    #     dict(
    #         source=img, # або місцевий шлях
    #         xref="x",
    #         yref="y",
    #         x=30,           # Початкова координата X (лівий край малюнка)
    #         y=260,          # Початкова координата Y (верхній край малюнка!)
    #         sizex=150,      # Ширина малюнка в мм
    #         sizey=400,      # Висота малюнка в мм
    #         sizing="contain",
    #         opacity=0.9,
    #         layer="below"
    #     )
    # )
    

    # --- 6. Налаштування полотна Plotly ---
    fig.update_layout(
        xaxis=dict(range=[-5, 215], showgrid=False, zeroline=False, visible=False, scaleanchor="y", scaleratio=1),
        yaxis=dict(range=[-5, 302], showgrid=False, zeroline=False, visible=False),
        width=650,
        height=920,
        showlegend=False,
        paper_bgcolor="white",
        plot_bgcolor="white",
        margin=dict(l=10, r=10, t=10, b=10)
    )

    return fig

# Відображення у Jupyter Notebook
fig = create_a4_frame_with_drawing()
fig.show()
```
