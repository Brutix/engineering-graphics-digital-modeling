# Тренажер: перетин поверхонь двох циліндрів

Розглядаємо два циліндри з перпендикулярними осями.

Циліндр 1 (вісь Z):  x² + y² = R1²

Циліндр 2 (вісь X):  (y − dy)² + (z − z0)² = R2²

Параметри R1, R2 задають радіуси; dy та z0 — зміщення осі другого циліндра.

На першому циліндрі точка має координати x=R1 cos(φ), y=R1 sin(φ). Підстановка в рівняння другого дає:

z = z0 ± √(R2² − (R1 sin(φ) − dy)²).

Дійсні значення кореня визначають точки спільної лінії поверхонь. Тренажер одночасно показує просторову модель та три проєкції.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from mpl_toolkits.mplot3d import Axes3D

def intersection(R1=2., R2=1.5, dy=0., z0=0.):
    phi=np.linspace(0,2*np.pi,180)
    zgrid=np.linspace(-3.5,3.5,80)
    xgrid=np.linspace(-3.5,3.5,80)

    # Циліндр 1: вертикальна вісь Z
    P1,Z1=np.meshgrid(phi,zgrid)
    X1,Y1=R1*np.cos(P1),R1*np.sin(P1)

    # Циліндр 2: горизонтальна вісь X
    P2,X2=np.meshgrid(phi,xgrid)
    Y2=dy+R2*np.cos(P2)
    Z2=z0+R2*np.sin(P2)

    t=np.linspace(0,2*np.pi,2400)
    x=R1*np.cos(t); y=R1*np.sin(t)
    rad=R2**2-(y-dy)**2
    valid=rad>=0
    root=np.sqrt(np.maximum(rad,0))
    zp=np.where(valid,z0+root,np.nan)
    zm=np.where(valid,z0-root,np.nan)

    fig=plt.figure(figsize=(14,9))
    ax=fig.add_subplot(221,projection="3d")
    axy=fig.add_subplot(222); axz=fig.add_subplot(223); ayz=fig.add_subplot(224)

    ax.plot_surface(X1,Y1,Z1,color="cornflowerblue",alpha=.28,linewidth=0)
    ax.plot_surface(X2,Y2,Z2,color="orange",alpha=.28,linewidth=0)
    ax.plot(x,y,zp,color="crimson",lw=2.8)
    ax.plot(x,y,zm,color="crimson",lw=2.8)
    ax.set(xlim=(-3.5,3.5),ylim=(-3.5,3.5),zlim=(-3.5,3.5),
           xlabel="X",ylabel="Y",zlabel="Z",title="Просторова модель")
    ax.set_box_aspect((1,1,1)); ax.view_init(elev=23,azim=-55)

    axy.plot(R1*np.cos(phi),R1*np.sin(phi),color="steelblue",alpha=.6)
    axy.plot(x,y,color="crimson",lw=2)
    axy.set(title="Горизонтальна проєкція X–Y",xlabel="X",ylabel="Y")
    axy.set_aspect("equal"); axy.grid(alpha=.25)

    axz.scatter(x[valid],zp[valid],s=2,color="crimson")
    axz.scatter(x[valid],zm[valid],s=2,color="crimson")
    axz.set(title="Фронтальна проєкція X–Z",xlabel="X",ylabel="Z")
    axz.grid(alpha=.25)

    ayz.scatter(y[valid],zp[valid],s=2,color="crimson")
    ayz.scatter(y[valid],zm[valid],s=2,color="crimson")
    ayz.set(title="Профільна проєкція Y–Z",xlabel="Y",ylabel="Z")
    ayz.grid(alpha=.25)

    if not np.any(valid):
        print("Поверхні не перетинаються.")
    else:
        print("Лінію перетину знайдено.")
        print(f"Кутових положень із дійсними точками: {valid.sum()} із {len(t)}")
        print(f"Діапазон X: {x[valid].min():.2f} … {x[valid].max():.2f}")
        print(f"Діапазон Y: {y[valid].min():.2f} … {y[valid].max():.2f}")
        print(f"Діапазон Z: {np.nanmin(zm):.2f} … {np.nanmax(zp):.2f}")
    plt.tight_layout(); plt.show()

interact(
    intersection,
    R1=FloatSlider(min=.5,max=3.,step=.1,value=2.,description="Радіус R1"),
    R2=FloatSlider(min=.5,max=3.,step=.1,value=1.5,description="Радіус R2"),
    dy=FloatSlider(min=-3.,max=3.,step=.1,value=0.,description="Зсув Y"),
    z0=FloatSlider(min=-2.,max=2.,step=.1,value=0.,description="Зсув Z")
)
```

## Практичні завдання

1. Встановіть R1 = R2. Дослідіть симетрію лінії перетину.
2. Зробіть R1 удвічі більшим за R2. Порівняйте проєкції.
3. Змінюйте dy. Знайдіть положення, за якого перетин зникає.
4. Змінюйте z0. Простежте, чи змінюється форма лінії, чи лише її положення.
5. Поясніть, чому точки лінії перетину мають одночасно задовольняти рівняння обох циліндрів.
