```python
import numpy as np
import plotly.graph_objects as go

t = np.linspace(0, 2*np.pi, 200)

x = np.cos(t)
y = np.sin(t)
z = np.zeros_like(t)

fig = go.Figure()

fig.add_trace(
    go.Scatter3d(
        x=x,
        y=y,
        z=z,
        mode="lines",
        name="Траєкторія"
    )
)

fig.update_layout(
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z"
    )
)

fig.show()
```
