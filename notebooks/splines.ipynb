```python
# ============================================================
# ПАРАМЕТРИ — ЗМІНЮЄ СТУДЕНТ
# ============================================================

R = 2.0          # радіус спряження
ANGLE = 60       # кут між прямими, градуси

L1 = 6.0         # довжина першої прямої
L2 = 5.0         # довжина другої прямої


# ============================================================
# АЛГОРИТМ — НЕ ЗМІНЮЄМО
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Переводимо кут у радіани
angle = np.radians(ANGLE)

# Вершина кута
B = np.array([0.0, 0.0])

# Напрямок першої прямої
d1 = np.array([-1.0, 0.0])

# Напрямок другої прямої
d2 = np.array([np.cos(angle), np.sin(angle)])

# ------------------------------------------------------------
# Знаходимо точки дотику
# ------------------------------------------------------------

t = R / np.tan(angle / 2)

P1 = B + d1 * t
P2 = B + d2 * t

# ------------------------------------------------------------
# Знаходимо центр дуги
# ------------------------------------------------------------

O = np.array([
    -t,
    R
])

# ------------------------------------------------------------
# Будуємо початкові прямі
# ------------------------------------------------------------

A = B + d1 * L1
C = B + d2 * L2

# ------------------------------------------------------------
# Будуємо дугу спряження
# ------------------------------------------------------------

a1 = np.arctan2(P1[1] - O[1], P1[0] - O[0])
a2 = np.arctan2(P2[1] - O[1], P2[0] - O[0])

angles = np.linspace(a1, a2, 100)

x = O[0] + R * np.cos(angles)
y = O[1] + R * np.sin(angles)


# ============================================================
# ВІЗУАЛІЗАЦІЯ
# ============================================================

plt.figure(figsize=(9, 6))

# Початкові прямі
plt.plot(
    [A[0], B[0]],
    [A[1], B[1]],
    linewidth=2,
    label="Пряма 1"
)

plt.plot(
    [B[0], C[0]],
    [B[1], C[1]],
    linewidth=2,
    label="Пряма 2"
)

# Спряження
plt.plot(
    x, y,
    linewidth=4,
    label=f"Спряження R = {R}"
)

# Точки дотику
plt.scatter(
    [P1[0], P2[0]],
    [P1[1], P2[1]],
    s=60,
    zorder=5,
    label="Точки дотику"
)

# Центр
plt.scatter(
    O[0], O[1],
    s=60,
    zorder=5,
    label="Центр дуги"
)

plt.axis("equal")
plt.grid(True)
plt.legend()

plt.title(
    f"Алгоритмічна побудова спряження\n"
    f"R = {R}, кут = {ANGLE}°"
)

plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```
