# Тренажер: точка на поверхні циліндра та її проекції

### Мета
Дослідити положення точки на бічній поверхні просторового тіла та побудову її трьох проекцій.

Циліндр задається параметрами:

- центр основи `C(Xc, Yc, Zc)`;
- радіус `R`;
- висота `H`.

Положення точки на бічній поверхні задається **двома параметрами**:

- `φ` — кут навколо осі циліндра;
- `Z` — висота точки.

Система автоматично обчислює координати точки:

**A(X, Y, Z)**

та її проекції:

- **A₁(X, Z)** — фронтальна;
- **A₂(X, Y)** — горизонтальна;
- **A₃(Y, Z)** — профільна.

> **NO-CAD:** студент не малює точку вручну. Він задає її положення параметрами, а система будує точку та всі її проекції.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Checkbox

def trainer(cx, cy, cz, radius, height, phi, z_point, show_links):
    # Точка ПРИВ'ЯЗАНА до поверхні циліндра
    a = np.deg2rad(phi)
    x = cx + radius*np.cos(a)
    y = cy + radius*np.sin(a)
    z = cz + z_point

    t = np.linspace(0, 2*np.pi, 120)
    zz = np.linspace(cz, cz+height, 50)
    T, Z = np.meshgrid(t, zz)
    X = cx + radius*np.cos(T)
    Y = cy + radius*np.sin(T)

    fig = plt.figure(figsize=(16,9))
    axf = fig.add_subplot(221)
    axp = fig.add_subplot(222)
    axh = fig.add_subplot(223)
    ax3 = fig.add_subplot(224, projection="3d")

    # ФРОНТАЛЬНА XZ
    axf.set_title("ФРОНТАЛЬНА ПРОЕКЦІЯ XZ", fontweight="bold")
    axf.set_xlabel("X"); axf.set_ylabel("Z"); axf.grid(alpha=.25)
    axf.set_aspect("equal", adjustable="box")
    axf.fill([cx-radius,cx+radius,cx+radius,cx-radius],
             [cz,cz,cz+height,cz+height], alpha=.08)
    axf.plot([cx-radius,cx+radius,cx+radius,cx-radius,cx-radius],
             [cz,cz,cz+height,cz+height,cz], linewidth=1.8)
    axf.scatter(x,z,s=110)
    axf.annotate("A₁",(x,z),xytext=(8,8),textcoords="offset points",
                 fontsize=13,fontweight="bold")

    # ПРОФІЛЬНА YZ
    axp.set_title("ПРОФІЛЬНА ПРОЕКЦІЯ YZ", fontweight="bold")
    axp.set_xlabel("Y"); axp.set_ylabel("Z"); axp.grid(alpha=.25)
    axp.set_aspect("equal", adjustable="box")
    axp.fill([cy-radius,cy+radius,cy+radius,cy-radius],
             [cz,cz,cz+height,cz+height], alpha=.08)
    axp.plot([cy-radius,cy+radius,cy+radius,cy-radius,cy-radius],
             [cz,cz,cz+height,cz+height,cz], linewidth=1.8)
    axp.invert_xaxis()
    axp.scatter(y,z,s=110)
    axp.annotate("A₃",(y,z),xytext=(8,8),textcoords="offset points",
                 fontsize=13,fontweight="bold")

    # ГОРИЗОНТАЛЬНА XY
    axh.set_title("ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ XY", fontweight="bold")
    axh.set_xlabel("X"); axh.set_ylabel("Y"); axh.grid(alpha=.25)
    axh.set_aspect("equal", adjustable="box")
    cx_circle = cx + radius*np.cos(t)
    cy_circle = cy + radius*np.sin(t)
    axh.fill(cx_circle,cy_circle,alpha=.08)
    axh.plot(cx_circle,cy_circle,linewidth=1.8)
    axh.scatter(x,y,s=110)
    axh.annotate("A₂",(x,y),xytext=(8,8),textcoords="offset points",
                 fontsize=13,fontweight="bold")

    # Проекційні зв'язки
    if show_links:
        axf.axvline(x,linestyle="--",linewidth=1,alpha=.35)
        axf.axhline(z,linestyle="--",linewidth=1,alpha=.35)
        axh.axvline(x,linestyle="--",linewidth=1,alpha=.35)
        axh.axhline(y,linestyle="--",linewidth=1,alpha=.35)
        axp.axvline(y,linestyle="--",linewidth=1,alpha=.35)
        axp.axhline(z,linestyle="--",linewidth=1,alpha=.35)

    # 3D
    ax3.set_title("3D: ТОЧКА ПРИВ'ЯЗАНА ДО ПОВЕРХНІ",fontweight="bold")
    ax3.plot_surface(X,Y,Z,alpha=.20,linewidth=0)
    ax3.plot(cx_circle,cy_circle,np.full_like(t,cz),linewidth=1.5)
    ax3.plot(cx_circle,cy_circle,np.full_like(t,cz+height),linewidth=1.5)
    ax3.plot([cx,cx],[cy,cy],[cz,cz+height],
             linestyle="--",linewidth=1.2,alpha=.6)

    # Радіус до точки: явно показує її належність поверхні
    ax3.plot([cx,x],[cy,y],[z,z],linestyle="--",linewidth=1.4)
    ax3.plot([x,x],[y,y],[cz,z],linestyle=":",linewidth=1.2)
    ax3.scatter(x,y,z,s=130)
    ax3.text(x,y,z,"  A",fontsize=13,fontweight="bold")

    ax3.set_xlabel("X"); ax3.set_ylabel("Y"); ax3.set_zlabel("Z")
    span=max(2*radius,height,4)
    ax3.set_xlim(cx-span/2,cx+span/2)
    ax3.set_ylim(cy-span/2,cy+span/2)
    ax3.set_zlim(cz,cz+span)

    fig.suptitle(
        f"A({x:.2f}, {y:.2f}, {z:.2f})   •   φ={phi:.0f}°   •   "
        f"R={radius:.1f}   •   Z точки={z_point:.1f}",
        fontsize=15,fontweight="bold")
    plt.tight_layout()
    plt.show()

interact(
    trainer,
    cx=FloatSlider(min=0,max=10,step=1,value=5,description="Циліндр X"),
    cy=FloatSlider(min=0,max=10,step=1,value=5,description="Циліндр Y"),
    cz=FloatSlider(min=0,max=5,step=1,value=0,description="Циліндр Z"),
    radius=FloatSlider(min=1,max=6,step=1,value=3,description="Радіус R"),
    height=FloatSlider(min=2,max=10,step=1,value=6,description="Висота H"),
    phi=FloatSlider(min=0,max=360,step=5,value=45,description="Кут φ"),
    z_point=FloatSlider(min=0,max=6,step=.5,value=4,description="Висота точки"),
    show_links=Checkbox(value=True,description="Показувати зв'язки")
)
```

### Головний висновок

Положення точки на поверхні циліндра можна задати параметрами **φ і Z**.

Система автоматично визначає:

**φ, Z → X,Y,Z → A₁,A₂,A₃**

Це і є перехід від просторового об'єкта до його параметричного цифрового опису.
