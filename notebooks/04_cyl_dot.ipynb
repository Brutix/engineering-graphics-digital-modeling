# Тренажер: точка на поверхні циліндра та її проекції

### Мета
Дослідити положення точки на бічній поверхні просторового тіла та побудову її трьох проекцій.

Циліндр задається параметрами:

- центр основи `C(Xc, Yc, Zc)`;
- радіус `R`;
- висота `H`.

Положення точки на бічній поверхні задається **двома параметрами**:

- `φ` — кут навколо осі циліндра;
- `Z` — висота точки.

Система автоматично обчислює координати точки:

**A(X, Y, Z)**

та її проекції:

- **A₁(X, Z)** — фронтальна;
- **A₂(X, Y)** — горизонтальна;
- **A₃(Y, Z)** — профільна.

> **NO-CAD:** студент не малює точку вручну. Він задає її положення параметрами, а система будує точку та всі її проекції.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Checkbox

def trainer(cx, cy, cz, radius, height, phi, point_z, show_links):
    # Координати точки на бічній поверхні циліндра
    angle = np.deg2rad(phi)
    x = cx + radius*np.cos(angle)
    y = cy + radius*np.sin(angle)
    z = point_z

    # Поверхня циліндра
    t = np.linspace(0, 2*np.pi, 100)
    zz = np.linspace(cz, cz+height, 40)
    T, Z = np.meshgrid(t, zz)
    X = cx + radius*np.cos(T)
    Y = cy + radius*np.sin(T)

    fig = plt.figure(figsize=(16, 9))

    axf = fig.add_subplot(221)
    axp = fig.add_subplot(222)
    axh = fig.add_subplot(223)
    ax3 = fig.add_subplot(224, projection="3d")

    # -------------------------------------------------
    # ФРОНТАЛЬНА ПРОЕКЦІЯ XZ
    # -------------------------------------------------
    axf.set_title("ФРОНТАЛЬНА ПРОЕКЦІЯ  XZ", fontweight="bold")
    axf.set_xlabel("X")
    axf.set_ylabel("Z")
    axf.grid(alpha=.25)
    axf.set_aspect("equal", adjustable="box")

    axf.fill(
        [cx-radius, cx+radius, cx+radius, cx-radius],
        [cz, cz, cz+height, cz+height],
        alpha=.10
    )
    axf.plot(
        [cx-radius, cx+radius, cx+radius, cx-radius, cx-radius],
        [cz, cz, cz+height, cz+height, cz],
        linewidth=1.8
    )

    axf.scatter(x, z, s=100)
    axf.annotate("A₁", (x,z), xytext=(8,8), textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # -------------------------------------------------
    # ПРОФІЛЬНА ПРОЕКЦІЯ YZ
    # -------------------------------------------------
    axp.set_title("ПРОФІЛЬНА ПРОЕКЦІЯ  YZ", fontweight="bold")
    axp.set_xlabel("Y")
    axp.set_ylabel("Z")
    axp.grid(alpha=.25)
    axp.set_aspect("equal", adjustable="box")
    axp.invert_xaxis()

    axp.fill(
        [cy-radius, cy+radius, cy+radius, cy-radius],
        [cz, cz, cz+height, cz+height],
        alpha=.10
    )
    axp.plot(
        [cy-radius, cy+radius, cy+radius, cy-radius, cy-radius],
        [cz, cz, cz+height, cz+height, cz],
        linewidth=1.8
    )

    axp.scatter(y, z, s=100)
    axp.annotate("A₃", (y,z), xytext=(8,8), textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # -------------------------------------------------
    # ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ XY
    # -------------------------------------------------
    axh.set_title("ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ  XY", fontweight="bold")
    axh.set_xlabel("X")
    axh.set_ylabel("Y")
    axh.grid(alpha=.25)
    axh.set_aspect("equal", adjustable="box")

    circle_x = cx + radius*np.cos(t)
    circle_y = cy + radius*np.sin(t)

    axh.fill(circle_x, circle_y, alpha=.10)
    axh.plot(circle_x, circle_y, linewidth=1.8)

    axh.scatter(x, y, s=100)
    axh.annotate("A₂", (x,y), xytext=(8,8), textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # -------------------------------------------------
    # ПРОЕКЦІЙНІ ЗВ'ЯЗКИ
    # -------------------------------------------------
    if show_links:
        axf.axvline(x, linestyle="--", linewidth=1, alpha=.35)
        axf.axhline(z, linestyle="--", linewidth=1, alpha=.35)

        axh.axvline(x, linestyle="--", linewidth=1, alpha=.35)
        axh.axhline(y, linestyle="--", linewidth=1, alpha=.35)

        axp.axvline(y, linestyle="--", linewidth=1, alpha=.35)
        axp.axhline(z, linestyle="--", linewidth=1, alpha=.35)

    # -------------------------------------------------
    # 3D-МОДЕЛЬ
    # -------------------------------------------------
    ax3.set_title("3D-МОДЕЛЬ: ТОЧКА НА ПОВЕРХНІ", fontweight="bold")
    ax3.plot_surface(X, Y, Z, alpha=.22, linewidth=0)

    # Основи циліндра
    ax3.plot(
        circle_x, circle_y, np.full_like(t, cz),
        linewidth=1.5
    )
    ax3.plot(
        circle_x, circle_y, np.full_like(t, cz+height),
        linewidth=1.5
    )

    # Вісь циліндра
    ax3.plot(
        [cx,cx], [cy,cy], [cz,cz+height],
        linestyle="--", linewidth=1.3, alpha=.6
    )

    # Точка A
    ax3.scatter(x, y, z, s=120)
    ax3.text(x, y, z, "  A", fontsize=13, fontweight="bold")

    # Радіус до точки — показує геометричний зв'язок
    ax3.plot(
        [cx,x], [cy,y], [z,z],
        linestyle="--", linewidth=1.2
    )

    # Висота до точки
    ax3.plot(
        [x,x], [y,y], [cz,z],
        linestyle=":", linewidth=1.2
    )

    ax3.set_xlabel("X")
    ax3.set_ylabel("Y")
    ax3.set_zlabel("Z")

    span = max(2*radius, height, 4)
    ax3.set_xlim(cx-span/2, cx+span/2)
    ax3.set_ylim(cy-span/2, cy+span/2)
    ax3.set_zlim(cz, cz+span)

    # -------------------------------------------------
    # Заголовок
    # -------------------------------------------------
    fig.suptitle(
        f"A({x:.2f}, {y:.2f}, {z:.2f})   •   φ = {phi:.0f}°   •   R = {radius:.1f}",
        fontsize=16,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()

interact(
    trainer,
    cx=FloatSlider(min=0, max=10, step=1, value=5, description="Центр X"),
    cy=FloatSlider(min=0, max=10, step=1, value=5, description="Центр Y"),
    cz=FloatSlider(min=0, max=5, step=1, value=0, description="Центр Z"),
    radius=FloatSlider(min=1, max=6, step=1, value=3, description="Радіус R"),
    height=FloatSlider(min=2, max=10, step=1, value=6, description="Висота H"),
    phi=FloatSlider(min=0, max=360, step=5, value=45, description="Кут φ"),
    point_z=FloatSlider(min=0, max=6, step=0.5, value=4, description="Висота точки Z"),
    show_links=Checkbox(value=True, description="Показувати зв'язки")
)
```

### Головний висновок

Положення точки на поверхні циліндра можна задати параметрами **φ і Z**.

Система автоматично визначає:

**φ, Z → X,Y,Z → A₁,A₂,A₃**

Це і є перехід від просторового об'єкта до його параметричного цифрового опису.
