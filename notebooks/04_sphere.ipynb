# Тренажер: точка на поверхні сфери та її проекції

### Мета

Перейти від циліндра і конуса до просторової поверхні, де положення точки визначається **двома кутами**.

Сфера задається:
- центром `C(Xc, Yc, Zc)`;
- радіусом `R`.

Положення точки A на поверхні задається:
- `φ` — азимутальний кут навколо осі Z;
- `θ` — кут від додатного напрямку осі Z.

Координати точки:

**X = Xc + R·sin(θ)·cos(φ)**

**Y = Yc + R·sin(θ)·sin(φ)**

**Z = Zc + R·cos(θ)**

Система автоматично будує:
- 3D-сферу;
- точку A;
- A₁ — фронтальну проекцію;
- A₂ — горизонтальну проекцію;
- A₃ — профільну проекцію.

> **NO-CAD:** студент задає параметри положення точки, а система сама визначає її координати та проекції.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Checkbox

def trainer(cx, cy, cz, radius, phi, theta, show_links):
    ph = np.deg2rad(phi)
    th = np.deg2rad(theta)

    # Координати точки A на сфері
    x = cx + radius*np.sin(th)*np.cos(ph)
    y = cy + radius*np.sin(th)*np.sin(ph)
    z = cz + radius*np.cos(th)

    # Сфера
    u = np.linspace(0, 2*np.pi, 100)
    v = np.linspace(0, np.pi, 60)
    U, V = np.meshgrid(u, v)

    X = cx + radius*np.sin(V)*np.cos(U)
    Y = cy + radius*np.sin(V)*np.sin(U)
    Z = cz + radius*np.cos(V)

    fig = plt.figure(figsize=(16, 9))

    axf = fig.add_subplot(221)
    axp = fig.add_subplot(222)
    axh = fig.add_subplot(223)
    ax3 = fig.add_subplot(224, projection="3d")

    # ---------- ФРОНТАЛЬНА ПРОЕКЦІЯ XZ ----------
    axf.set_title("ФРОНТАЛЬНА ПРОЕКЦІЯ XZ", fontweight="bold")
    axf.set_xlabel("X")
    axf.set_ylabel("Z")
    axf.grid(alpha=.25)
    axf.set_aspect("equal", adjustable="box")

    t = np.linspace(0, 2*np.pi, 200)
    axf.plot(cx + radius*np.cos(t),
             cz + radius*np.sin(t), linewidth=1.8)
    axf.fill(cx + radius*np.cos(t),
             cz + radius*np.sin(t), alpha=.08)

    axf.scatter(x, z, s=110)
    axf.annotate("A₁", (x,z), xytext=(8,8),
                 textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # ---------- ПРОФІЛЬНА ПРОЕКЦІЯ YZ ----------
    axp.set_title("ПРОФІЛЬНА ПРОЕКЦІЯ YZ", fontweight="bold")
    axp.set_xlabel("Y")
    axp.set_ylabel("Z")
    axp.grid(alpha=.25)
    axp.set_aspect("equal", adjustable="box")

    axp.plot(cy + radius*np.cos(t),
             cz + radius*np.sin(t), linewidth=1.8)
    axp.fill(cy + radius*np.cos(t),
             cz + radius*np.sin(t), alpha=.08)

    axp.scatter(y, z, s=110)
    axp.annotate("A₃", (y,z), xytext=(8,8),
                 textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # ---------- ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ XY ----------
    axh.set_title("ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ XY", fontweight="bold")
    axh.set_xlabel("X")
    axh.set_ylabel("Y")
    axh.grid(alpha=.25)
    axh.set_aspect("equal", adjustable="box")

    axh.plot(cx + radius*np.cos(t),
             cy + radius*np.sin(t), linewidth=1.8)
    axh.fill(cx + radius*np.cos(t),
             cy + radius*np.sin(t), alpha=.08)

    axh.scatter(x, y, s=110)
    axh.annotate("A₂", (x,y), xytext=(8,8),
                 textcoords="offset points",
                 fontsize=13, fontweight="bold")

    # ---------- ПРОЕКЦІЙНІ ЗВ'ЯЗКИ ----------
    if show_links:
        axf.axvline(x, linestyle="--", linewidth=1, alpha=.35)
        axf.axhline(z, linestyle="--", linewidth=1, alpha=.35)

        axh.axvline(x, linestyle="--", linewidth=1, alpha=.35)
        axh.axhline(y, linestyle="--", linewidth=1, alpha=.35)

        axp.axvline(y, linestyle="--", linewidth=1, alpha=.35)
        axp.axhline(z, linestyle="--", linewidth=1, alpha=.35)

    # ---------- 3D-МОДЕЛЬ ----------
    ax3.set_title("3D-МОДЕЛЬ: ТОЧКА НА СФЕРІ", fontweight="bold")
    ax3.plot_surface(X, Y, Z, alpha=.20, linewidth=0)

    # Екватор
    ax3.plot(cx + radius*np.cos(t),
             cy + radius*np.sin(t),
             np.full_like(t, cz), linewidth=1.5)

    # Меридіан через точку
    mer_v = np.linspace(0, np.pi, 150)
    mer_x = cx + radius*np.sin(mer_v)*np.cos(ph)
    mer_y = cy + radius*np.sin(mer_v)*np.sin(ph)
    mer_z = cz + radius*np.cos(mer_v)

    ax3.plot(mer_x, mer_y, mer_z,
             linestyle="--", linewidth=1.2, alpha=.7)

    # Паралель через точку
    par_u = np.linspace(0, 2*np.pi, 150)
    par_r = radius*np.sin(th)
    par_x = cx + par_r*np.cos(par_u)
    par_y = cy + par_r*np.sin(par_u)
    par_z = np.full_like(par_u, z)

    ax3.plot(par_x, par_y, par_z,
             linestyle="--", linewidth=1.2, alpha=.7)

    # Радіус до точки
    ax3.plot([cx,x], [cy,y], [cz,z],
             linestyle=":", linewidth=1.5)

    ax3.scatter(x, y, z, s=130)
    ax3.text(x, y, z, "  A", fontsize=13, fontweight="bold")

    ax3.scatter(cx, cy, cz, s=40)

    ax3.set_xlabel("X")
    ax3.set_ylabel("Y")
    ax3.set_zlabel("Z")

    span = 2*radius
    ax3.set_xlim(cx-span/2, cx+span/2)
    ax3.set_ylim(cy-span/2, cy+span/2)
    ax3.set_zlim(cz-span/2, cz+span/2)

    fig.suptitle(
        f"A({x:.2f}, {y:.2f}, {z:.2f})   •   "
        f"R = {radius:.1f}   •   φ = {phi:.0f}°   •   θ = {theta:.0f}°",
        fontsize=15, fontweight="bold"
    )

    plt.tight_layout()
    plt.show()

interact(
    trainer,
    cx=FloatSlider(min=0, max=10, step=1, value=5, description="Центр X"),
    cy=FloatSlider(min=0, max=10, step=1, value=5, description="Центр Y"),
    cz=FloatSlider(min=0, max=10, step=1, value=5, description="Центр Z"),
    radius=FloatSlider(min=1, max=6, step=1, value=4, description="Радіус R"),
    phi=FloatSlider(min=0, max=360, step=5, value=45, description="Кут φ"),
    theta=FloatSlider(min=0, max=180, step=5, value=60, description="Кут θ"),
    show_links=Checkbox(value=True, description="Показувати зв'язки")
)
```

### Головний висновок

Для сфери положення точки на поверхні визначається двома кутами:

**R, φ, θ → X,Y,Z → A₁,A₂,A₃**

Отже, одна і та сама просторова точка має координати та три пов'язані проекції.
