# Інтерактивне спряження двох прямих

Геометрично правильна модель: центр дуги лежить на бісектрисі кута, а дуга дотична до обох прямих.

Студент керує тільки **R** та **кутом**.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def draw_fillet(R=2.0, ANGLE=90):
    alpha = np.radians(ANGLE)
    B = np.array([0.0, 0.0])

    # Сторони кута
    d1 = np.array([-1.0, 0.0])
    d2 = np.array([np.cos(alpha), np.sin(alpha)])

    # Бісектриса внутрішнього кута
    bisector = d1 + d2
    bisector = bisector / np.linalg.norm(bisector)

    # Відстань від вершини до центра та до точок дотику
    BO = R / np.sin(alpha / 2)
    t = R / np.tan(alpha / 2)

    # Центр та точки дотику
    O = B + bisector * BO
    P1 = B + d1 * t
    P2 = B + d2 * t

    # Довжина відображуваних прямих
    L = max(6.0, t + 2.0)
    A = B + d1 * L
    C = B + d2 * L

    # Кути радіусів OP1 та OP2
    a1 = np.arctan2(P1[1]-O[1], P1[0]-O[0])
    a2 = np.arctan2(P2[1]-O[1], P2[0]-O[0])

    # Вибираємо дугу всередині кута
    delta = (a2 - a1) % (2*np.pi)
    if delta > np.pi:
        a1, a2 = a2, a1 + 2*np.pi
    else:
        a2 = a1 + delta

    aa = np.linspace(a1, a2, 200)
    x = O[0] + R*np.cos(aa)
    y = O[1] + R*np.sin(aa)

    # Графік
    fig, ax = plt.subplots(figsize=(10, 7))

    # Початковий кут
    ax.plot([A[0], B[0]], [A[1], B[1]], '--', linewidth=1.3,
            label='Початкові лінії')
    ax.plot([B[0], C[0]], [B[1], C[1]], '--', linewidth=1.3)

    # Обрізані лінії
    ax.plot([A[0], P1[0]], [A[1], P1[1]], linewidth=3)
    ax.plot([P2[0], C[0]], [P2[1], C[1]], linewidth=3)

    # Справжня дуга спряження
    ax.plot(x, y, linewidth=5, label=f'Спряження R = {R:.1f}')

    # Радіуси та бісектриса
    ax.plot([O[0], P1[0]], [O[1], P1[1]], ':', linewidth=1.5)
    ax.plot([O[0], P2[0]], [O[1], P2[1]], ':', linewidth=1.5)
    E = B + bisector*BO*1.25
    ax.plot([B[0], E[0]], [B[1], E[1]], '-.', linewidth=1.2,
            label='Бісектриса')

    # Точки
    ax.scatter([P1[0], P2[0]], [P1[1], P2[1]], s=65, zorder=5,
               label='Точки дотику')
    ax.scatter(O[0], O[1], s=65, zorder=5, label='Центр O')

    ax.text(B[0]+0.12, B[1]-0.35, 'B — вершина', fontsize=11)
    ax.text(P1[0]-0.15, P1[1]-0.35, 'P₁', fontsize=11)
    ax.text(P2[0]+0.12, P2[1]+0.08, 'P₂', fontsize=11)
    ax.text(O[0]+0.12, O[1]+0.12, 'O', fontsize=11)

    # Перевірка: відстані від O до обох прямих = R
    d_to_1 = abs(np.cross(d1, O-B))
    d_to_2 = abs(np.cross(d2, O-B))

    ax.text(0.02, 0.98,
            f'R = {R:.2f}\nкут = {ANGLE}°\n'
            f'BO = {BO:.2f}\n'
            f'd(O, пряма 1) = {d_to_1:.2f}\n'
            f'd(O, пряма 2) = {d_to_2:.2f}',
            transform=ax.transAxes, va='top', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.5', alpha=0.9))

    ax.set_aspect('equal', adjustable='box')
    ax.grid(True, alpha=0.25)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_title('Алгоритмічна побудова спряження двох прямих')
    ax.legend(loc='best')
    plt.show()

interact(
    draw_fillet,
    R=FloatSlider(value=2.0, min=0.5, max=5.0, step=0.1,
                  description='Радіус R:', continuous_update=True),
    ANGLE=IntSlider(value=90, min=20, max=160, step=5,
                    description='Кут:', continuous_update=True)
);
```

## Що змінюється автоматично

- центр **O**;
- точки дотику **P₁, P₂**;
- положення дуги;
- частини прямих після умовного `Trim`.

При зміні **R** або кута спряження залишається дотичним до обох ліній.
