# Комп'ютерна графіка — тренажер «Три проекції»

**NO-CAD:** студент задає просторовий об'єкт координатами, а Notebook автоматично будує його графічне представлення.

Модель: **A(x,y,z), B(x,y,z), C(x,y,z)** → площина ABC → фронтальна XZ, горизонтальна XY, профільна YZ та допоміжний 3D-вигляд.

```python
import numpy as np\nimport matplotlib.pyplot as plt\nfrom ipywidgets import interact, FloatSlider, Checkbox, Dropdown, HBox, VBox, HTML\nfrom mpl_toolkits.mplot3d import Axes3D\n\nplt.rcParams['figure.figsize'] = (15, 9)\nplt.rcParams['font.size'] = 11\n\nPOINT_COLORS = {'A': 'tab:red', 'B': 'tab:blue', 'C': 'tab:green'}\n\ndef make_points(Ax, Ay, Az, Bx, By, Bz, Cx, Cy, Cz):\n    return np.array([[Ax, Ay, Az], [Bx, By, Bz], [Cx, Cy, Cz]], dtype=float)\n\ndef plane_type(P):\n    A, B, C = P\n    u = B - A\n    v = C - A\n    n = np.cross(u, v)\n    if np.linalg.norm(n) < 1e-9:\n        return 'Площина не визначена: точки A, B, C колінеарні.'\n    if np.isclose(P[:, 2], P[0, 2]).all():\n        return 'Горизонтальна площина (z = const)'\n    if np.isclose(P[:, 1], P[0, 1]).all():\n        return 'Фронтальна площина (y = const)'\n    if np.isclose(P[:, 0], P[0, 0]).all():\n        return 'Профільна площина (x = const)'\n    return 'Площина загального положення'\n\ndef limits(P):\n    lo = P.min(axis=0); hi = P.max(axis=0)\n    span = np.maximum(hi-lo, 2)\n    return lo-span*0.35, hi+span*0.35\n\ndef draw_projection(ax, P, dims, title, show_lines):\n    i, j = dims\n    ax.set_title(title, fontweight='bold')\n    ax.set_xlabel(['X','Y','Z'][i]); ax.set_ylabel(['X','Y','Z'][j])\n    lo, hi = limits(P)\n    ax.set_xlim(lo[i], hi[i]); ax.set_ylim(lo[j], hi[j])\n    ax.set_aspect('equal', adjustable='box')\n    ax.grid(True, alpha=0.25)\n    Q = P[:, [i,j]]\n    ax.fill(Q[:,0], Q[:,1], alpha=0.10, edgecolor='black')\n    ax.plot(Q[[0,1,2,0],0], Q[[0,1,2,0],1], color='black', lw=1.5)\n    for k, name in enumerate(['A','B','C']):\n        ax.scatter(Q[k,0], Q[k,1], s=70, color=POINT_COLORS[name], zorder=3)\n        ax.annotate(name, (Q[k,0], Q[k,1]), xytext=(7,7), textcoords='offset points', fontweight='bold')\n        if show_lines:\n            ax.axvline(Q[k,0], color=POINT_COLORS[name], ls='--', alpha=.18)\n            ax.axhline(Q[k,1], color=POINT_COLORS[name], ls='--', alpha=.18)\n\ndef draw_3d(ax, P):\n    ax.set_title('Допоміжний 3D-вигляд', fontweight='bold')\n    ax.plot(P[[0,1,2,0],0], P[[0,1,2,0],1], P[[0,1,2,0],2], color='black', lw=1.5)\n    ax.plot_trisurf(P[:,0], P[:,1], P[:,2], alpha=.12)\n    for k, name in enumerate(['A','B','C']):\n        ax.scatter(*P[k], s=70, color=POINT_COLORS[name])\n        ax.text(P[k,0], P[k,1], P[k,2], '  '+name, fontweight='bold')\n    lo, hi = limits(P)\n    ax.set_xlim(lo[0],hi[0]); ax.set_ylim(lo[1],hi[1]); ax.set_zlim(lo[2],hi[2])\n    ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')\n    ax.view_init(elev=24, azim=-58)\n\ndef show_trainer(Ax=2, Ay=2, Az=5, Bx=8, By=2, Bz=5, Cx=5, Cy=7, Cz=5, show_lines=True):\n    P = make_points(Ax,Ay,Az,Bx,By,Bz,Cx,Cy,Cz)\n    fig = plt.figure(figsize=(15,9))\n    ax1 = fig.add_subplot(221)\n    ax2 = fig.add_subplot(222)\n    ax3 = fig.add_subplot(223)\n    ax4 = fig.add_subplot(224, projection='3d')\n    draw_projection(ax1,P,(0,2),'Фронтальна проекція XZ',show_lines)\n    draw_projection(ax2,P,(1,2),'Профільна проекція YZ',show_lines)\n    draw_projection(ax3,P,(0,1),'Горизонтальна проекція XY',show_lines)\n    draw_3d(ax4,P)\n    fig.suptitle(plane_type(P), fontsize=15, fontweight='bold')\n    plt.tight_layout()\n    plt.show()\n\ninteract(show_trainer,\n    Ax=FloatSlider(min=0,max=10,step=1,value=2,description='A: X'),\n    Ay=FloatSlider(min=0,max=10,step=1,value=2,description='A: Y'),\n    Az=FloatSlider(min=0,max=10,step=1,value=5,description='A: Z'),\n    Bx=FloatSlider(min=0,max=10,step=1,value=8,description='B: X'),\n    By=FloatSlider(min=0,max=10,step=1,value=2,description='B: Y'),\n    Bz=FloatSlider(min=0,max=10,step=1,value=5,description='B: Z'),\n    Cx=FloatSlider(min=0,max=10,step=1,value=5,description='C: X'),\n    Cy=FloatSlider(min=0,max=10,step=1,value=7,description='C: Y'),\n    Cz=FloatSlider(min=0,max=10,step=1,value=5,description='C: Z'),\n    show_lines=Checkbox(value=True,description='Проекційні лінії')\n);
```
