# Тренажер: параметричне просторове тіло — циліндр

**Мета:** задавати просторовий об'єкт параметрами та бачити автоматичну зміну його 3D-моделі і трьох проекцій.

**Циліндр = положення + радіус + висота**

NO-CAD: ми не малюємо об'єкт вручну — описуємо його, а система будує геометричне представлення.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Checkbox

def trainer(cx, cy, cz, radius, height, show_axis):
    t = np.linspace(0, 2*np.pi, 100)
    z = np.linspace(cz, cz+height, 2)
    T, Z = np.meshgrid(t, z)
    X = cx + radius*np.cos(T)
    Y = cy + radius*np.sin(T)

    fig = plt.figure(figsize=(16, 9))
    axf = fig.add_subplot(221)
    axp = fig.add_subplot(222)
    axh = fig.add_subplot(223)
    ax3 = fig.add_subplot(224, projection="3d")

    # Фронтальна XZ
    axf.set_title("ФРОНТАЛЬНА ПРОЕКЦІЯ XZ", fontweight="bold")
    axf.set_xlabel("X"); axf.set_ylabel("Z"); axf.grid(alpha=.25)
    axf.set_aspect("equal", adjustable="box")
    axf.fill([cx-radius,cx+radius,cx+radius,cx-radius],
             [cz,cz,cz+height,cz+height], alpha=.12)
    axf.plot([cx-radius,cx+radius,cx+radius,cx-radius,cx-radius],
             [cz,cz,cz+height,cz+height,cz])

    # Профільна YZ
    axp.set_title("ПРОФІЛЬНА ПРОЕКЦІЯ YZ", fontweight="bold")
    axp.set_xlabel("Y"); axp.set_ylabel("Z"); axp.grid(alpha=.25)
    axp.set_aspect("equal", adjustable="box")
    axp.fill([cy-radius,cy+radius,cy+radius,cy-radius],
             [cz,cz,cz+height,cz+height], alpha=.12)
    axp.plot([cy-radius,cy+radius,cy+radius,cy-radius,cy-radius],
             [cz,cz,cz+height,cz+height,cz])

    # Горизонтальна XY
    axh.set_title("ГОРИЗОНТАЛЬНА ПРОЕКЦІЯ XY", fontweight="bold")
    axh.set_xlabel("X"); axh.set_ylabel("Y"); axh.grid(alpha=.25)
    axh.set_aspect("equal", adjustable="box")
    axh.fill(cx+radius*np.cos(t), cy+radius*np.sin(t), alpha=.12)
    axh.plot(cx+radius*np.cos(t), cy+radius*np.sin(t))

    # 3D
    ax3.set_title("3D-МОДЕЛЬ", fontweight="bold")
    ax3.plot_surface(X, Y, Z, alpha=.28)
    ax3.plot(cx+radius*np.cos(t), cy+radius*np.sin(t),
             np.full_like(t,cz), linewidth=1.5)
    ax3.plot(cx+radius*np.cos(t), cy+radius*np.sin(t),
             np.full_like(t,cz+height), linewidth=1.5)
    if show_axis:
        ax3.plot([cx,cx],[cy,cy],[cz,cz+height],linestyle="--")

    ax3.set_xlabel("X"); ax3.set_ylabel("Y"); ax3.set_zlabel("Z")

    # Єдині масштаби для 2D-проекцій
    m = max(radius*1.4, height*.65, 2)
    axf.set_xlim(cx-m,cx+m); axf.set_ylim(cz-m,cz+height+m*.25)
    axp.set_xlim(cy-m,cy+m); axp.set_ylim(cz-m,cz+height+m*.25)
    axh.set_xlim(cx-m,cx+m); axh.set_ylim(cy-m,cy+m)

    fig.suptitle(
        f"Циліндр: центр ({cx:.0f}, {cy:.0f}, {cz:.0f})  •  R = {radius:.0f}  •  H = {height:.0f}",
        fontsize=16, fontweight="bold")
    plt.tight_layout()
    plt.show()

interact(
    trainer,
    cx=FloatSlider(min=0,max=10,step=1,value=5,description="Центр X"),
    cy=FloatSlider(min=0,max=10,step=1,value=5,description="Центр Y"),
    cz=FloatSlider(min=0,max=10,step=1,value=0,description="Центр Z"),
    radius=FloatSlider(min=1,max=6,step=1,value=3,description="Радіус R"),
    height=FloatSlider(min=1,max=10,step=1,value=6,description="Висота H"),
    show_axis=Checkbox(value=True,description="Показувати вісь")
)
```
